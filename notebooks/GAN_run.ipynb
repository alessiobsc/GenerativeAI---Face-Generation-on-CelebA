```python
import torch

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Uso device:", DEVICE)
```

```python
DRIVE_WEIGHTS_DIR = "/content/drive/MyDrive/Generative AI Project/gan/weights"
CHECKPOINT_NAME   = "cgan_generator_only.pt"
LATENT_SIZE = 128
NUM_CLASSES = 8

N_PER_CLASS = 4
SEED        = 123

OUT_DIR = "/content/results"

CHECKPOINT_PATH = f"{DRIVE_WEIGHTS_DIR}/{CHECKPOINT_NAME}"
print("Checkpoint:", CHECKPOINT_PATH)
print("Output dir:", OUT_DIR)
```

```python
import torch

raw = torch.load(CHECKPOINT_PATH, map_location="cpu", weights_only=False)

SCRIPT_CHECKPOINT_PATH = CHECKPOINT_PATH
print("Formato compatibile, uso:", SCRIPT_CHECKPOINT_PATH)
```

```python
import subprocess

cmd = [
    "python", "generate_cgan.py",
    "--checkpoint", SCRIPT_CHECKPOINT_PATH,
    "--out_dir", OUT_DIR,
    "--n_per_class", str(N_PER_CLASS),
    "--seed", str(SEED),
    "--device", DEVICE
]
print("Comando:", " ".join(cmd))
subprocess.run(cmd, check=True)
```
